```python
num_1 = float( input('digite o primeiro número: ').replace(',', '.')) # aguardar primeiro número
num_2 = float( input('digite o segundo numero: ').replace(',', '.')) # aguardar segundo número

operação = input('''Digite a operação desejada
+ para adição, - para subtração, * para multiplicação, / for divisão
''')

if operação == '+':
    print('{} + {} =' . format(num_1, num_2))
    print(num_1 + num_2)
elif operação == '-':
    print('{} - {} =' . format(num_1, num_2))
    print(num_1 - num_2)
elif operação == '*':
    print('{} * {} =' . format(num_1, num_2))
    print(num_1 * num_2)
elif operação == '/':
    print('{} / {} =' . format(num_1, num_2))
    print(num_1 / num_2)
```
